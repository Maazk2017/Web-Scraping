A robots.txt file uses a standard set of text-based directives known as the Robots Exclusion Protocol (REP). Web crawlers (like search engine bots or AI scrapers) read these lines from top to bottom to determine where they are allowed to go on a website.

1. User-agent
What it does: Specifies which crawler or bot the subsequent rules apply to.

How it works: Every bot has a unique token or name. You can write rules for a specific bot or use an asterisk (*) as a wildcard to target all crawlers universally.

Examples:

User-agent: * (Applies to all automated crawlers)

User-agent: Googlebot (Applies strictly to Google's web crawler)

User-agent: GPTBot (Applies strictly to OpenAI's web crawler)

2. Disallow
What it does: Instructs crawlers not to access specific files, directories, or paths on the website.

How it works: The path following Disallow: tells the bot which URLs to skip. If the path is left completely blank (Disallow:), it means "do not disallow anything" (allow full access for that user-agent). A forward slash alone (Disallow: /) blocks the entire website.

Examples:

Disallow: /admin/ (Blocks bots from crawling anything inside the /admin/ folder)

Disallow: /private-page.html (Blocks bots from crawling a specific file)

Disallow: / (Blocks bots from crawling the entire site)

3. Allow
What it does: Explicitly grants permission to crawl a specific subdirectory or file that would otherwise be blocked by a broader Disallow rule.

How it works: Because crawlers process rules based on specificity (usually matching the longest path prefix), an Allow directive can override a parent Disallow rule.

Example:

Plaintext
Disallow: /folder/
Allow: /folder/public-file.html
(This blocks bots from accessing everything in /folder/, except for public-file.html.)

4. Sitemap
What it does: Points crawlers directly to the website's XML sitemap(s), which list all important pages, videos, and files available for indexing.

How it works: Unlike the other commands, Sitemap is not tied to a specific User-agent block; it is usually placed at the very top or bottom of the robots.txt file using a fully qualified URL.

Example:

Sitemap: [https://example.com/sitemap.xml](https://example.com/sitemap.xml)

5. Crawl-delay (Optional / Non-Standard)
What it does: Instructs bots to wait a specified number of seconds between consecutive requests to prevent server overload.

How it works: While historically supported by some search engines (like Bing and Yandex), major search engines like Google ignore Crawl-delay in robots.txt, preferring that site owners manage crawl rates directly via their webmaster console tools.

Example:

Crawl-delay: 10 (Asks the bot to wait 10 seconds between page requests)

Putting It All Together: A Sample robots.txt File
Plaintext
# Block AI training bots completely from the entire site
User-agent: GPTBot
Disallow: /

# Rules for all other standard search crawlers
User-agent: *
Disallow: /cart/
Disallow: /checkout/
Disallow: /user-account/
Allow: /cart/help-faq

# Provide the sitemap location for index discovery
Sitemap: https://example.com/sitemap.xml

```python
import requests
from urllib.parse import urlparse

def fetch_robots_txt(url, output_filename="robots.txt"):
    if not url.startswith(('http://', 'https://')):
        url = 'https://' + url
        
    parsed_url = urlparse(url)
    base_domain = parsed_url.netloc or parsed_url.path
    robots_url = f"https://{base_domain}/robots.txt"

    # Use a polite User-Agent header
    headers = {
        'User-Agent': 'ComplianceAuditBot/1.0 (Educational/Reconnaissance Research)'
    }

    print(f"[*] Fetching: {robots_url}")

    try:
        resp = requests.get(robots_url, headers=headers, timeout=10)
        
        if resp.status_code == 200:
            with open(output_filename, 'w', encoding='utf-8') as f:
                f.write(resp.text)
            print(f"[+] Success! Saved robots.txt to: {output_filename}")

        elif resp.status_code == 404:
            print("[-] Notice: robots.txt not found (404 Not Found). The site may not have one configured.")
        else:
            print(f"[-] Warning: Received HTTP status code {resp.status_code}")
            
    except requests.exceptions.RequestException as e:
        print(f"[-] Connection error occurred: {e}")

fetch_robots_txt("https://amazon.com", output_filename="robots.txt")
```
